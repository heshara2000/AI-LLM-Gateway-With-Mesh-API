```python
import os

from getpass import getpass
from dotenv import load_dotenv
load_dotenv()
```

```python
from meshapi import meshapi
MESHAPI_TOKEN=os.getenv("MESH_API_KEY") or getpass("MeshAPI token (rsk_...):")

client=MeshAPI(
    base_url=os.getenv("MESHAPI_BASE_URL", "https://api.meshapi.ai"),
    token = MESHAPI_TOKEN
)

print("Client ready)
```

```python
len(client.models.list())
```

```python
ALL_MODELS = {m.id: m  for m in client.models.list()}

print(f"{len(ALL_MODELS)} MODELS AVAIBLE THROUGH THIS KEY")
```
